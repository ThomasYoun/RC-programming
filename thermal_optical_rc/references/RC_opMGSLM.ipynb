```python
import torch
import numpy as np


class OpticalSLMRC:
    """
    Optical SLM reservoir-computing simulation.

    Optical path used here:
        reservoir state + input image
        -> vertical SLM phase concatenation
        -> pad/crop to 128 x 128 optical plane
        -> objective 1 pupil transfer
        -> fixed implicit scattering operator
        -> optional heater phase perturbation
        -> objective 2 pupil transfer
        -> CCD intensity
        -> camera downsampling to reservoir state
        -> linear readout

    Important convention:
        The implicit W_opt is mathematically a 128*128 by 128*128 operator,
        but it is never materialized as a dense matrix. The fixed random
        scattering medium is represented by fixed random phase screens.
    """

    def __init__(
        self,
        n_inputs: int = 1,
        n_reservoir: int | None = None,
        n_outputs: int = 1,
        ridge_alpha: float = 1e-6,
        leak_rate: float = 1.0,
        
        input_shape   = (64, 64),
        image_shape   = (64, 64),
        slm_shape     = (128, 128),
        optical_shape = (128, 128),
        input_scaling:   float = 0.8,    # amplitude of W_in
        spectral_radius: float = 0.8,
        objective1_radius: float = 0.25,
        objective2_radius: float = 0.25,
        n_scattering_layers: int = 1,
        
        use_heater: bool = False,
        heater_strength: float = 0.0,
        heater_center_i=None,
        heater_center_j=None,
        heater_radius: int = 12,
        heater_sigma=None,
        slm_noise_type=None,
        slm_noise_sigma: float = 0.3,
        camera_noise_type=None,
        camera_noise_sigma: float = 0.3,
        seed: int = 42,
        device: str = "cuda" if torch.cuda.is_available() else "cpu",
    ):
        self.n_inputs = int(n_inputs)
        self.n_outputs = int(n_outputs)
        self.ridge_alpha = ridge_alpha
        self.leak_rate = leak_rate
        self.input_scaling = input_scaling
        self.spectral_radius = float(spectral_radius)
        self.device = torch.device(device)
        self.H_gain = torch.tensor(self.spectral_radius, device=self.device)
        self.seed = seed

        self.input_shape = tuple(int(v) for v in input_shape)
        self.image_shape = tuple(int(v) for v in image_shape)
        self.slm_shape = tuple(int(v) for v in slm_shape)
        self.optical_shape = tuple(int(v) for v in optical_shape)

        self.H, self.W = self.image_shape
        self.N = self.H * self.W
        self.n_reservoir = self.N if n_reservoir is None else int(n_reservoir)
        if self.n_reservoir != self.N:
            raise ValueError(
                f"n_reservoir ({self.n_reservoir}) must equal image_shape[0]*image_shape[1] ({self.N})."
            )

        self.N_slm = self.slm_shape[0] * self.slm_shape[1]
        self.N_optical = self.optical_shape[0] * self.optical_shape[1]

        self.objective1_radius = objective1_radius
        self.objective2_radius = objective2_radius
        self.n_scattering_layers = int(n_scattering_layers)

        self.use_heater = use_heater
        self.heater_strength = heater_strength
        self.heater_center_i = heater_center_i
        self.heater_center_j = heater_center_j
        self.heater_radius = heater_radius
        self.heater_sigma = heater_sigma

        self.slm_noise_type = slm_noise_type
        self.slm_noise_sigma = slm_noise_sigma
        self.camera_noise_type = camera_noise_type
        self.camera_noise_sigma = camera_noise_sigma

        self.generator = torch.Generator(device=self.device)
        self.generator.manual_seed(seed)
        torch.manual_seed(seed)

        self.W_out = None
        self.snapshot = None
        self._init_weights()
        self.reset_state()

    # ------------------------------------------------------------------
    # Fixed optical components
    # ------------------------------------------------------------------
    def _init_weights(self):
        """Initialize fixed masks/operators once. Do not call this per step."""
        self.input_mask = torch.rand(
            self.input_shape,
            device=self.device,
            generator=self.generator,
        )

        self.pupil1 = self.make_pupil(self.optical_shape, self.objective1_radius)
        self.pupil2 = self.make_pupil(self.optical_shape, self.objective2_radius)

        self.scatter_weight_pre = []
        self.scatter_weight_fourier = []
        for _ in range(self.n_scattering_layers):
            pre = self.make_complex_amplitude_matrix(self.optical_shape)
            fourier = self.make_complex_amplitude_matrix(self.optical_shape)
            
            self.scatter_weight_pre.append(pre)
            self.scatter_weight_fourier.append(fourier)

        self.heater_profile = self.build_circular_decay_mask(
            shape=self.optical_shape,
            center_i=self.heater_center_i,
            center_j=self.heater_center_j,
            radius=self.heater_radius,
            sigma=self.heater_sigma,
        )

    def reset_state(self):
        self.r_img = torch.zeros(self.image_shape, device=self.device)
        self.snapshot = None

    def make_pupil(self, shape, radius=0.4):
        """Circular objective pupil in normalized Fourier coordinates."""
        H, W = shape
        yy, xx = torch.meshgrid(
            torch.linspace(-1.0, 1.0, H, device=self.device),
            torch.linspace(-1.0, 1.0, W, device=self.device),
            indexing="ij",
        )
        return ((xx ** 2 + yy ** 2) <= radius ** 2).float()
        
    def make_complex_amplitude_matrix(self, shape):
        mask = (
            (torch.rand(shape, device=self.device, generator=self.generator) * 2 - 1)
            + 1j * (torch.rand(shape, device=self.device, generator=self.generator) * 2 - 1)
        )

        mask = mask * self.input_scaling/ (mask.abs().pow(2).mean().sqrt() + 1e-12)
        return mask

    def build_circular_decay_mask(self, shape, center_i=None, center_j=None, radius=12, sigma=None):
        """Heater phase profile on the 128 x 128 optical plane."""
        H, W = shape
        if center_i is None:
            center_i = H // 2
        if center_j is None:
            center_j = W // 2
        if sigma is None:
            sigma = max(radius / 2, 1.0)

        ii, jj = torch.meshgrid(
            torch.arange(H, device=self.device),
            torch.arange(W, device=self.device),
            indexing="ij",
        )
        dist2 = (ii - center_i) ** 2 + (jj - center_j) ** 2
        decay = torch.exp(-dist2 / (2 * sigma ** 2))
        aperture = (dist2 <= radius ** 2).float()
        return decay * aperture

    # ------------------------------------------------------------------
    # Utility transforms
    # ------------------------------------------------------------------
    def pad_or_crop_to_shape(self, field, target_shape):
        """Center pad/crop a 2D field to target_shape."""
        target_h, target_w = target_shape
        src_h, src_w = field.shape
        out = torch.zeros(target_shape, dtype=field.dtype, device=field.device)

        copy_h = min(src_h, target_h)
        copy_w = min(src_w, target_w)

        src_i0 = max((src_h - copy_h) // 2, 0)
        src_j0 = max((src_w - copy_w) // 2, 0)
        dst_i0 = max((target_h - copy_h) // 2, 0)
        dst_j0 = max((target_w - copy_w) // 2, 0)

        out[dst_i0:dst_i0 + copy_h, dst_j0:dst_j0 + copy_w] = field[
            src_i0:src_i0 + copy_h,
            src_j0:src_j0 + copy_w,
        ]
        return out

    def normalize01(self, x):
        return (x - x.min()) / (x.max() - x.min() + 1e-8)

    def camera_to_reservoir(self, intensity):
        """Downsample 128 x 128 camera intensity to H x W reservoir state."""
        intensity = intensity[None, None, :, :]
        return torch.nn.functional.adaptive_avg_pool2d(
            intensity,
            output_size=self.image_shape,
        )[0, 0]

    # ------------------------------------------------------------------
    # Noise
    # ------------------------------------------------------------------
    def add_gaussian_noise(self, x, sigma=0.05):
        return torch.clamp(x + sigma * torch.randn_like(x), 0.0, 1.0)

    def add_poisson_noise(self, x, peak=255):
        x_scaled = torch.clamp(x, 0.0, 1.0) * peak
        return torch.clamp(torch.poisson(x_scaled) / peak, 0.0, 1.0)

    def add_speckle_noise(self, x, sigma=0.1):
        return torch.clamp(x + x * sigma * torch.randn_like(x), 0.0, 1.0)

    def add_salt_pepper_noise(self, x, prob=0.01):
        noisy = x.clone()
        rand = torch.rand_like(x)
        noisy[rand < prob / 2] = 0.0
        noisy[rand > 1 - prob / 2] = 1.0
        return noisy

    def add_noise(self, x, noise_type=None, sigma=0.1):
        if noise_type is None:
            return x
        if noise_type == "gaussian":
            return self.add_gaussian_noise(x, sigma=sigma)
        if noise_type == "poisson":
            peak = max(10, min(255, int(1.0 / (sigma + 1e-6))))
            return self.add_poisson_noise(x, peak=peak)
        if noise_type == "speckle":
            return self.add_speckle_noise(x, sigma=sigma)
        if noise_type == "salt_pepper":
            return self.add_salt_pepper_noise(x, prob=sigma)
        raise ValueError(f"Unknown noise type: {noise_type}")

    # ------------------------------------------------------------------
    # SLM encoding
    # ------------------------------------------------------------------
    def encode_u_to_image(self, u_t: torch.Tensor) -> torch.Tensor:
        """Scalar u(t) in [0, 1] -> fixed-mask binary input image."""
        u_t = torch.as_tensor(u_t, dtype=torch.float32, device=self.device).flatten()[0]
        return (self.input_mask < u_t.clamp(0.0, 1.0)).float()

    def build_slm_phase_field(self, u_t: torch.Tensor) -> dict:
        """
        Vertical SLM concatenation.

        reservoir state is padded/cropped to the upper half of slm_shape;
        input image is padded/cropped to the lower half. The result is phase
        encoded as exp(i*pi*x).
        """
        input_img = torch.round(self.encode_u_to_image(u_t) * 255) / 255
  
        state_img = self.r_img
        if torch.is_complex(state_img):
            state_img = (torch.angle(state_img) + torch.pi) / (2 * torch.pi)
        else:
            state_img = self.normalize01(state_img)
        state_img = torch.round(state_img * 255) / 255

        half_h = self.slm_shape[0] // 2
        upper = self.pad_or_crop_to_shape(state_img, (half_h, self.slm_shape[1]))
        lower = self.pad_or_crop_to_shape(input_img, (self.slm_shape[0] - half_h, self.slm_shape[1]))
        slm_phase_amp = torch.cat([upper, lower], dim=0)

        if self.slm_noise_type is not None:
            slm_phase_amp = self.add_noise(
                slm_phase_amp,
                noise_type=self.slm_noise_type,
                sigma=self.slm_noise_sigma,
            )
        slm_phase_amp = torch.clamp(slm_phase_amp, 0.0, 1.0)
        slm_field = torch.exp(1j * torch.pi * slm_phase_amp)

        return {
            "slm_field": slm_field,
            "input_img": input_img,
            "state_img": state_img,
            "slm_phase_amp": slm_phase_amp,
            "slm_phase_img": torch.angle(slm_field),
        }

    # ------------------------------------------------------------------
    # Optical propagation
    # ------------------------------------------------------------------
    def apply_objective(self, field, pupil):
        """FFT -> Fourier-space pupil -> inverse FFT."""
        E_k = torch.fft.fftshift(torch.fft.fft2(field))
        E_k = E_k * pupil
        return torch.fft.ifft2(torch.fft.ifftshift(E_k))

    def apply_heater(self, field):
        if not self.use_heater or self.heater_strength == 0:
            return field
        heater_phase = (self.heater_strength * self.heater_profile)*torch.exp(1j * self.heater_strength * self.heater_profile)
        return field * heater_phase

    def apply_H_unscaled(self, field):
        """
        Fixed implicit scattering operator.

        This represents a structured 16384 x 16384 optical transfer operator
        on a 128 x 128 plane without storing the dense matrix.
        """ 
        E = field.to(torch.cfloat)
        for scatter_phase_pre, scatter_phase_fourier in zip(self.scatter_weight_pre, self.scatter_weight_fourier):
            E = E * scatter_phase_pre
            E = self.apply_heater(E)
            E = torch.fft.fft2(E, norm="ortho")
            E = E * scatter_phase_fourier
            E = torch.fft.ifft2(E,norm="ortho")
        return E
        
    def apply_H(self, field):
        """
        Scaled implicit H(theta) @ field.
        """
        return self.H_gain * self.apply_H_unscaled(field)
        
    def apply_scattering(self, field):
        """
        Fixed implicit scattering operator on a 2D optical field.
        """
        return self.apply_H(field)
    

    def calculate_camera_intensity(self, slm_field):
        """Full optical path from SLM plane to 128 x 128 CCD intensity."""
        E_slm = self.pad_or_crop_to_shape(slm_field, self.optical_shape)
        E_focus = self.apply_objective(E_slm, self.pupil1)
        E_scattered = self.apply_scattering(E_focus)
        E_camera = self.apply_objective(E_scattered, self.pupil2)
        intensity = torch.abs(E_camera) ** 2
        return self.normalize01(intensity)

    # Backward-compatible name used in older notebook cells.
    def calculate_speckle_torch(self, field, img_shape=None, out_shape=None):
        field = self.pad_or_crop_to_shape(field, self.optical_shape)
        intensity = self.calculate_camera_intensity(field)
        if out_shape is not None:
            intensity = torch.nn.functional.adaptive_avg_pool2d(
                intensity[None, None, :, :],
                output_size=out_shape,
            )[0, 0]
        return intensity
        
    @torch.no_grad()
    def estimate_spectral_radius(self, n_iter=100, eps=1e-3):
        """
        Estimate spectral radius of the implicit unscaled H.

        For the current phase-only scattering with ortho FFT, rho is theoretically 1.
        This function is useful if you later add amplitude loss/gain.
        """
        z = torch.randn(self.optical_shape, device=self.device)
        z = z + 1j * torch.randn(self.optical_shape, device=self.device)
        z = z / (z.norm() + eps)

        lam = torch.tensor(1.0, device=self.device)

        for _ in range(n_iter):
            Hz = self.apply_H_unscaled(z)
            lam = torch.vdot(z.reshape(-1), Hz.reshape(-1))
            z = Hz / (Hz.norm() + eps)
        return lam.abs()


    @torch.no_grad()
    def calibrate_spectral_radius(self, n_iter=100):
        rho = self.estimate_spectral_radius(n_iter=n_iter)

        gain = self.spectral_radius / (rho + 1e-3)

        if not torch.is_tensor(self.H_gain):
            self.H_gain = torch.tensor(gain, device=self.device)
        else:
            self.H_gain.copy_(gain.detach())

        return rho
    
    # ------------------------------------------------------------------
    # Reservoir dynamics
    # ------------------------------------------------------------------
    def _step(self, u_t: torch.Tensor, save=False, t=None) -> torch.Tensor:
        alpha = self.leak_rate

        data = self.build_slm_phase_field(u_t)
        intensity = self.calculate_camera_intensity(data["slm_field"])

        if self.camera_noise_type is not None:
            intensity = self.add_noise(
                intensity,
                noise_type=self.camera_noise_type,
                sigma=self.camera_noise_sigma,
            )

        intensity_res = self.camera_to_reservoir(intensity)
        r_new = torch.sqrt(intensity_res + 1e-8)

        state_t = self.r_img.clone()
        self.r_img = (1 - alpha) * self.r_img + alpha * r_new

        if save:
            self.snapshot = {
                "t": t,
                "u": float(torch.as_tensor(u_t).detach().cpu().squeeze()),
                "use_heater": self.use_heater,
                "heater_strength": self.heater_strength,
                "heater_profile": self.heater_profile.detach().cpu(),
                "input_img": data["input_img"].detach().cpu(),
                "state_t": state_t.detach().cpu(),
                "slm_phase_img": data["slm_phase_img"].detach().cpu(),
                "camera_intensity": intensity.detach().cpu(),
                "state_t1": self.r_img.detach().cpu(),
            }

        return self.r_img.clone(), self.snapshot

    def _run(self, u: torch.Tensor, save_t=None) -> torch.Tensor:
        u = u.to(self.device).flatten()
        T = u.shape[0]
        states = torch.zeros(T, self.N, device=self.device)
        self.snapshot = None

        for t in range(T):
            save_now = save_t is not None and t == save_t
            r_img, _ = self._step(u[t], save=save_now, t=t)
            states[t] = r_img.flatten()
            if save_now:
                self.plot_snapshot()
        return states

    def fit(self, u_train: torch.Tensor, y_train: torch.Tensor, washout: int = 200, snapshot_t=None):
        print(f"Running reservoir over {u_train.shape[0]} training steps...")
        u_train = u_train.to(self.device)
        y_train = y_train.to(self.device)
        if y_train.ndim == 1:
            y_train = y_train.reshape(-1, 1)

        X = self._run(u_train, save_t=snapshot_t)
        X_fit = X[washout:]
        y_fit = y_train[washout:]

        I = torch.eye(X_fit.shape[1], device=self.device)
        A = X_fit.T @ X_fit + self.ridge_alpha * I
        A = A + 1e-6 * torch.eye(A.shape[0], device=self.device)
        b = X_fit.T @ y_fit
        self.W_out = torch.linalg.solve(A, b)

        y_out = X_fit @ self.W_out
        mse = ((y_out - y_fit) ** 2).mean().item()
        print(f"Training MSE: {mse:.2e}")
        return y_out

    def predict(self, u_test: torch.Tensor) -> torch.Tensor:
        assert self.W_out is not None, "Call .fit() before .predict()."
        X_test = self._run(u_test)
        return X_test @ self.W_out

    # ------------------------------------------------------------------
    # Visualization
    # ------------------------------------------------------------------
    def plot_snapshot(self):
        import matplotlib.pyplot as plt

        if self.snapshot is None:
            raise ValueError("snapshot is None. Run fit(..., snapshot_t=...) first.")

        snap = self.snapshot
        fig, axes = plt.subplots(2, 3, figsize=(15, 8))
        axes = axes.ravel()

        axes[0].imshow(snap["input_img"], cmap="gray", aspect="equal")
        axes[0].set_title("Input image")

        axes[1].imshow(snap["state_t"], cmap="viridis", aspect="equal")
        axes[1].set_title("State r(t)")

        im2 = axes[2].imshow(snap["slm_phase_img"], cmap="twilight", aspect="auto", vmin=-np.pi, vmax=np.pi)
        fig.colorbar(im2, ax=axes[2], fraction=0.046, pad=0.04)
        axes[2].set_title("SLM phase field")

        im3 = axes[3].imshow(snap["camera_intensity"], cmap="gray", aspect="equal")
        fig.colorbar(im3, ax=axes[3], fraction=0.046, pad=0.04)
        axes[3].set_title("CCD intensity 128x128")

        axes[4].imshow(snap["state_t1"], cmap="viridis", aspect="equal")
        axes[4].set_title("State r(t+1)")

        axes[5].imshow(snap["state_t1"] - snap["state_t"], cmap="bwr", aspect="equal")
        axes[5].set_title("State change")

        for ax in axes:
            ax.axis("off")
        plt.tight_layout()
        plt.show()

    def plot_heater_area(self, ax=None):
        import matplotlib.pyplot as plt

        if ax is None:
            _, ax = plt.subplots(figsize=(5, 5))
        im = ax.imshow(self.heater_profile.detach().cpu(), cmap="hot", origin="lower")
        ax.set_title("Heater profile on optical plane")
        ax.set_xlabel("x")
        ax.set_ylabel("y")
        ax.figure.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
        return self.heater_profile
```

```python
## Generation of Mackey Glass Data
n_total = 4000
washout = 200
n_train = 2000
n_test  = 1000

series = mackey_glass(n_total, tau=17)
series_min = series.min()
series_max = series.max()

series_norm = (series - series_min) / (series_max - series_min + 1e-8)

# one-step-ahead task: input u(t) -> target y = u(t+1)
u = torch.tensor(series_norm[:-1], dtype=torch.float32).unsqueeze(-1)  # (T, 1)
y = torch.tensor(series_norm[1:],  dtype=torch.float32).unsqueeze(-1)  # (T, 1)

u_train, y_train = u[:n_train],               y[:n_train]
u_test,  y_test  = u[n_train:n_train+n_test], y[n_train:n_train+n_test]

model = OpticalSLMRC(
    input_shape     = (64,64),
    n_reservoir     = 64*64,    # try 256 or 1024 and observe the effect
    image_shape     = (64, 64),
    slm_shape       = (128, 128),
    optical_shape   = (128, 128),
    n_outputs       = 1,
    #spectral_radius = 0.8,   # near 1 = long memory (slow echo decay)
    #sparsity        = 0.2,
    #input_scaling   = 0.7,
    ridge_alpha     = 0.02,
    leak_rate       = 0.8,
    use_heater=True,
    slm_noise_type="speckle",
    camera_noise_type="speckle",
)


# without heater or noise
y_out = model.fit(u_train, y_train, washout=200, snapshot_t=500)
snap=model.snapshot
y_pred = model.predict(u_test)

mse = ((y_pred.cpu() - y_test) ** 2).mean().item()
print("Test MSE:", mse)
print("Example true u(t+1):", float(y_test[0]))
print("Example predicted u(t+1):", float(y_pred[0].cpu()))

#plt.savefig('MG results SLM1.png', dpi=400)
```

```python
# previous results

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

# =========================
# Convert tensors to numpy
# =========================
def to_numpy(x):
    import numpy as np
    if isinstance(x, np.ndarray):
        return x.flatten()
    else:
        return x.detach().cpu().numpy().flatten()
y_train_pred_np = to_numpy(y_out)
y_pred_np = to_numpy(y_pred)
y_test_np = to_numpy(y_test)

series_np = to_numpy(series_norm)

test_rmse = np.sqrt(np.mean((y_pred_np - y_test_np) ** 2))

# =========================
# Visualize
# =========================
fig = plt.figure(figsize=(14, 9), constrained_layout=True)
gs = GridSpec(3, 2, figure=fig, hspace=0.48, wspace=0.32)

# panel 1: full series with train/test bands
ax1 = fig.add_subplot(gs[0, :])

ax1.plot(
    series_norm,
    color="k",
    lw=0.7,
    alpha=0.85,
    label="Mackey-Glass"
)

# y_out starts after washout
ax1.plot(
    np.arange(washout, n_train),
    y_train_pred_np,
    color="b",
    lw=0.7,
    alpha=0.85,
    label="Train prediction"
)

ax1.plot(
    np.arange(n_train, n_train + n_test),
    y_pred_np,
    color="r",
    lw=0.7,
    alpha=0.85,
    label="Test prediction"
)

ax1.set_xlim(0, 3000)
ax1.axvspan(0, n_train, alpha=0.05, color="blue")
ax1.axvspan(n_train, n_train + n_test, alpha=0.05, color="red")
ax1.set_title("Mackey-Glass Chaotic Time Series", fontsize=12, pad=8)
ax1.legend(framealpha=0.2, fontsize=9)

for sp in ax1.spines.values():
    sp.set_color("#333")


# panel 2: prediction vs truth
ax2 = fig.add_subplot(gs[1, :])

n_show = 300

ax2.plot(
    y_test_np[:n_show],
    color="k",
    lw=1.6,
    label="Ground truth",
    alpha=0.9
)

ax2.plot(
    y_pred_np[:n_show],
    color="r",
    lw=1.2,
    linestyle="--",
    label="Optical RC prediction",
    alpha=0.9
)

ax2.set_title(
    f"One-step-ahead Prediction on Test Set "
    f"(first {n_show} steps, RMSE = {test_rmse:.5f})",
    fontsize=12,
    pad=8
)

ax2.legend(framealpha=0.2, fontsize=9)

for sp in ax2.spines.values():
    sp.set_color("#333")


# panel 3: absolute error
ax3 = fig.add_subplot(gs[2, 0])

abs_error = np.abs(y_pred_np - y_test_np)

ax3.plot(
    abs_error,
    color="g",
    lw=0.7,
    alpha=0.85
)

ax3.set_title("Absolute Error (full test set)", fontsize=11, pad=8)
ax3.set_xlabel("Time step")
ax3.set_ylabel("|prediction - truth|")

for sp in ax3.spines.values():
    sp.set_color("#333")


# panel 4: scatter truth vs predicted
ax4 = fig.add_subplot(gs[2, 1])

ax4.scatter(
    y_test_np,
    y_pred_np,
    alpha=0.25,
    s=4,
    color="r"
)

lims = [
    min(y_test_np.min(), y_pred_np.min()),
    max(y_test_np.max(), y_pred_np.max())
]

ax4.plot(
    lims,
    lims,
    color="k",
    lw=1.2,
    linestyle="--",
    label="Perfect"
)

ax4.set_xlim(lims)
ax4.set_ylim(lims)
ax4.set_xlabel("Ground truth", fontsize=9)
ax4.set_ylabel("Predicted", fontsize=9)
ax4.set_title("Scatter: Truth vs Predicted", fontsize=11, pad=8)
ax4.legend(framealpha=0.2, fontsize=9)

for sp in ax4.spines.values():
    sp.set_color("#333")


plt.tight_layout()
#plt.savefig('MG results SLM2.png', dpi=400)
plt.show()
```
